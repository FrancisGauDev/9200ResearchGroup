```python
from pathlib import Path

import pandas as pd

hydro_dir = Path("Data") / "Hydro"

def load_hydro_year(year):
    header_row = 1 if year <= 2021 else 3
    data = pd.read_excel(
        hydro_dir / f"BalancingAuthorityLoad{year}.xls",
        header=header_row,
    )
    data.columns = ["date", "hour_ending", "load"]
    data["date"] = pd.to_datetime(data["date"])
    return data

hydro_2025 = load_hydro_year(2025)
hydro_2020_2024 = pd.concat(
    [load_hydro_year(year) for year in range(2020, 2025)],
    ignore_index=True,
)

print("2025:", hydro_2025.shape)
print("2020-2024:", hydro_2020_2024.shape)
```
